# Export a Chaldene Workflow to a [PWD File](https://github.com/pythonworkflow/python-workflow-definition)

## Overview

Chaldene allows you to export any workflow as a [Python Workflow Definition (PWD)](https://github.com/pythonworkflow/python-workflow-definition) file directly from the visual editor. Right-click anywhere on the canvas to open the context menu and select **Export as PWD**.

![Export as PWD](export2pwd.gif)

The exported `workflow.json` file can then be loaded and executed using the `python_workflow_definition` package, as shown below.

```python
from python_workflow_definition.pyiron_base import load_workflow_json
workflow_json_filename = "workflow.json"
delayed_object_lst = load_workflow_json(file_name=workflow_json_filename)
delayed_object_lst[-1].pull()
```
