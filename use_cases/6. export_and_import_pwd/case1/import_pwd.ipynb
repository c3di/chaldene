# Import a [PWD File](https://github.com/pythonworkflow/python-workflow-definition) into Chaldene

Chaldene can import an existing [Python Workflow Definition (PWD)](https://github.com/pythonworkflow/python-workflow-definition) and rebuild it as a visual workflow. Right-click anywhere on the canvas to open the context menu and select **Import PWD**, then pick the `workflow.json` file — the matching `workflow.py` is loaded automatically from the same folder.

Chaldene creates a visual node for each function defined in `workflow.py` and connects them according to `workflow.json`, so the imported workflow can be inspected, edited, and executed directly in the editor.
